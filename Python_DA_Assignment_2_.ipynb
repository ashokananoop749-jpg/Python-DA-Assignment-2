```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
# Load the 'taxis' dataset
df = sns.load_dataset("taxis")
```

```python
missing_columns = df.columns[df.isnull().any()].tolist()
print(missing_columns)
```

```python
df["pickup"] = pd.to_datetime(df["pickup"], errors="coerce")
df["dropoff"] = pd.to_datetime(df["dropoff"], errors="coerce")
```

```python
numerical_columns = df.select_dtypes(include=np.number).columns
for col in missing_columns:
    if df[col].isnull().sum() > 0:
        mode_value = df[col].mode()
        if not mode_value.empty:
            df[col] = df[col].fillna(mode_value[0])
```

```python
df.isnull().sum()
```

```python
line_data = df.sort_values("pickup")

plt.figure(figsize=(12, 6))
plt.plot(
    line_data["pickup"],
    line_data["fare"],
    color="blue",
    linewidth=1
)

plt.title("Fare Over Time")
plt.xlabel("Pickup Time")
plt.ylabel("Fare")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
borough_fare=df.groupby('pickup_borough')['fare'].sum()
plt.figure(figsize=(10,6))
plt.xlabel('Pickup Borough')
plt.ylabel('Total Fare')
plt.title('Total Fare by Pickup Borough')
plt.bar(borough_fare.index,borough_fare.values)

plt.show()
```

```python
payment_counts = df["payment"].value_counts()

plt.figure(figsize=(8, 8))

plt.pie(
payment_counts,
labels=payment_counts.index,
autopct="%1.1f%%",
startangle=90
)

plt.title("Distribution of Trips by Payment Method")
plt.axis("equal")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(
df["distance"],
bins=30,
color="green",
edgecolor="black"
)

plt.title("Distribution of Taxi Trip Distance")
plt.xlabel("Distance")
plt.ylabel("Number of Trips")


plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

df.boxplot(
column="tip",
by="pickup_borough",
figsize=(10, 6)
)

plt.title("Tip Distribution by Pickup Borough")
plt.suptitle("")
plt.xlabel("Pickup Borough")
plt.ylabel("Tip Amount")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
sns.countplot(
data=df,
x="pickup_borough",
color="blue"
)

plt.title("Number of Trips by Pickup Borough")
plt.xlabel("Pickup Borough")
plt.ylabel("Number of Trips")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
sns.scatterplot(x="distance",
y="fare",
hue="pickup_borough",data=df)

plt.title("Relationship Between Distance and Fare")
plt.xlabel("Distance")
plt.ylabel("Fare")
plt.legend(title="Pickup Borough")
plt.figure(figsize=(10, 6))
plt.tight_layout()
plt.show()
```

```python
corr = df[["distance", "fare", "tip", "tolls", "total"]].corr()
sns.heatmap(corr,annot=True)
plt.show()
```
